# 04 - Differential Privacy Implementation & Noise Calibration

Explores Laplace Mechanism sensitivity calibration, noise distribution, and bounded privacy budgets.


```python
import sys, os
sys.path.append('..')
import numpy as np
import matplotlib.pyplot as plt
from privacy.differential_privacy import LaplaceMechanism

epsilons = [0.1, 0.5, 1.0]
plt.figure(figsize=(10, 5))
for eps in epsilons:
    dp = LaplaceMechanism(epsilon=eps, sensitivity=1.0, seed=42)
    samples = [dp.draw_noise() for _ in range(5000)]
    plt.hist(samples, bins=60, alpha=0.5, density=True, label=f'epsilon={eps} (scale={1/eps:.1f})')
plt.title('Laplace Noise Distribution by Epsilon')
plt.xlabel('Noise Value')
plt.ylabel('Density')
plt.legend()
plt.show()
```
