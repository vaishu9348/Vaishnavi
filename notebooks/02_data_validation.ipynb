# 02 - Data Validation & PII Sanitization Audit

Verifies zero PII leakage, validates event schema integrity, and audits timestamp bucketization.


```python
import sys, os
sys.path.append('..')
import pandas as pd
from privacy.anonymization import PROHIBITED_PII_FIELDS

df = pd.read_csv('../data/synthetic/interaction_events.csv')
print('Total events:', len(df))

# Audit columns against prohibited PII
pii_found = [col for col in df.columns if col.lower() in PROHIBITED_PII_FIELDS]
print('Prohibited PII columns found:', pii_found)
assert len(pii_found) == 0, 'PII Leakage Detected!'
print('Zero PII compliance: PASSED ✅')
```
